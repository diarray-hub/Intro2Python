4.5 Écrivez un programme qui calcule le volume d’un parallélépipède rectangle dont sont fournis
au départ la largeur, la hauteur et la profondeur.

L'aire totale d'un parallélépipède rectangle (ou pavé droit) se calcule avec la formule :

$$ \text{Aire} = 2 \times (L \times l + L \times h + l \times h) $$

Et son volume se calcule avec la formule :

$$ \text{Volume} = L \times l \times h $$


```python
# code du volume d'un parallélépipède
largeur =float(input("Donner la largeur du parallélépipède"))
longeur =float(input("Donner la hauteur du parallélépipède"))
profondeur = float(input("Donner la profondeur de cette parallélopipède"))
volume = largeur*longeur*profondeur

print(volume)
```

Explication du code :

    input() : permet de demander à l'utilisateur de saisir une valeur.

    float() : convertit la chaîne de caractères saisie en nombre décimal (pour accepter des valeurs comme 3.5).

    Formule : le volume d'un parallélépipède rectangle est donné par :

    V=largeur×hauteur×profondeurV=largeur×hauteur×profondeur

    print(f"...") : affiche le résultat en utilisant une f-string pour insérer la valeur du volume.

4.6 Écrivez un programme qui convertit un nombre entier de secondes fourni au départ en un
nombre d’années, de mois, de jours, de minutes et de secondes (utilisez l’opérateur modulo : %
).

```python
#code convertisseur de nombre seconde 

#nbre_sec = int(input("Donner un nombre en seconde"))
nbre_sec = 1000000000
#le nombre de secondes initiales est stocké dans la variable total_sec_initial
total_sec_initial = nbre_sec
# conversion du nombre de secondes en heures, jours et mois
nbre_sec_minute = 60
nbre_sec_heur = nbre_sec_minute* 60
nbre_sec_jour = nbre_sec_heur *24
nbre_sec_mois = nbre_sec_jour * 30
nbre_sec_annee = nbre_sec_mois * 12 #ci on a as pris compte des different type d'année

#calcule de la conversion 
annees = total_sec_initial // nbre_sec_annee
#on estrait le nouveaux nombre de secondes restant après le calcul des années
total_sec_initial = total_sec_initial % nbre_sec_annee

#on calcule le nombre de mois
mois = total_sec_initial // nbre_sec_mois
#on estrait le nouveaux nombre de secondes restant après le calcul des mois
total_sec_initial = total_sec_initial % nbre_sec_mois
#on calcule le nombre de jours
jours = total_sec_initial // nbre_sec_jour
#on estrait le nouveaux nombre de secondes restant après le calcul des jours
total_sec_initial = total_sec_initial % nbre_sec_jour
#on calcule le nombre d'heures
heures = total_sec_initial // nbre_sec_heur 
#on estrait le nouveaux nombre de secondes restant après le calcul des heures 
total_sec_initial = total_sec_initial % nbre_sec_heur


print(f'le nombre d\'années est {annees} ans')
print(f'le nombre de mois est {mois} mois')
print(f'le nombre de jours est {jours} jours')
print(f'le nombre d\'heures est {heures} heures')
print(f'le nombre de secondes est {total_sec_initial} secondes')

```
