**ecrire un programme fibunacci avec une fonction**

```python
def fibunnacci(n):
    a, b = 0, 1
    #mise à jour du programme fibunnacci
    fic = [a]
    while (b<n):
        a, b = b, a + b
        fic.append(a)
    return fic
    
fibunnacci(1000)
```

```python
fibunnacci(10)
```

exercise pour faire une fonction trier qui nous permet de voire dans un classeur (dictionaire qui comprennent deux dictonnaire (positive et negatives)) et en fonction des données de classé les élement dns leur emplacement correspondant .

```python
classeur = {
  "positive":[],
  "negatives":[]
  
}
```

```python
#ici la fonction trier vas permettre de voir
def trier(classeur,nombre):
  # tier un nombre positive et négative
  # vasvoir si un nombre est positive le classe la value positive(liste) dans le dictionnaire classeur 
  if nombre >0:
    classeur['positive'].append(nombre)
  else : 
    classeur['negatives'].append(nombre)
  return classeur 
      
```
