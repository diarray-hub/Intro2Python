## Affichage formaté!
On commence par une première notion très largement repandue, comment obtenir perfomer le meilleur en terme d'affichage en python! On l'a vue la méthode print est la méthode standard de sortie en python, mais en réalité on peut bien plus que **print(x)**. Il arrive souvent qu'on ait besoin d'afficher un message d'une manière particulière comme incorporer des infos variables à un message par défaut, il y'a plusieurs manière de faire ce genre d'affichage des méthodes dérivées des autres langages basée sur la syntax avec **l'operateur (%)** aux **formatted strings**. On va voir certains de ces méthodes des moins populaires aux plus populaire!

## Pseudo formattage en utilisant les propriétés des strs et de la fonction print!
Avant de parler des méthodes de formattage en elles même, parlons de quelques syntaxs qu'on a déjà vu ici mais qu'on ne peut pas rééllement considérer comme du formattage! Par exemple sachant qu'on peut utiliser l'operateur "+" pour concatener deux strs on peut utiliser cela pour intégrer des strings entre eux au moment de l'affichage, on peut aussi modifier les propriétés d'un afficage en modifiant la valeur des arguments de la fonctions print!

```python
num = 666
name = "Abdou"
solde = 12345789.0
# Par défaut l'argument sep de la fonction print va permettre d'afficher ces arguments séparés par des espaces
# Il est important de noter que pour la fonction print, chaque valeur affichée ici est un argument positionnel à part
print("Compte numéro:", num, "; propriétaire:", name, "; solde:", solde)
# On peut également utiliser '+' pour concatener les strings comme dit précédemment
# Cette voici du point de vue de la fonction il n'y a plus qu'un seul argument
print("Compte numéro: " + str(num) + "; propriétaire: " + name + "; solde: " + str(solde))
```

## Affichage formatée avec l'opérateur "%"
Une méthode rélativement peu utilisé depuis Python3 mais plutôt populaire avant et considéré comme la seule méthode d'affichage formatée dans d'autres langages comme le C par exemple, on peut utiliser l'opérateur "%" suivi d'un suffix variables en fonctions du type de l'objet qu'on souhaite afficher pour intégrer des ints (%d), floats (%f) et strs (%s)

```python
print("Compte numéro: %d"%num)
formatting = (num, name, solde)
print("Compte numéro: %d; propriétaire: %s; solde: %.1f" %formatting)
# C'est le rôle de l'opérateur % losrqu'il est de cette manière avec les strings
# Notez qu'il est important d'avoir les abstractions "%x" dans le string à formater
# On peut même les utiliser comme opérateur de convertions
convert = "Compte numéro: %f" %num
print(convert)
# Disons que je veux limiter cet affichage deux chiffres après la virgule
print("Compte numéro %.1f" %num)
```

## Strings methods!
Les strs eux même en python fournissent des methodes pour faire de l'affichage formatée, jettons un petit coup d'oeil aux methodes **format, center, ljust and rjust**!
La méthode **format** fournit un autre moyen d'intégrer des objets de différents type dans un string, les autres permettent d'ajuster la position d'un message par rapport à un prompt!

```python
print("Compte numéro: {}; propriétaire: {}; solde: {}".format(num, name, solde))
# On peut configurer un systems de clés pour cette méthode
print("Compte numéro: {key1}; propriétaire: {key}; solde: {key0}".format(key0=solde, key1=num, key=name))
# Positionner un message proportionnellement à un prompt
message = "Abdou"
print(message.center(11, "$"))
print(message.ljust(10, "$"))
print(message.rjust(10, "$"))
```

## Formatted strings!
Les **F-strings ou String formatées** sont un type de str permattant de rendre la syntax des méthode comme format, center et etc plus concise et compréhensible, disponible depuis la version 3.6 de python, cette méthode est largement devenu la plus populaire dès lors grâce à sa syntax plus naturelle pour les débutants!<br/>
Pour créer un formatted string (string formattées) il suffit de le prefixer du caratère "f". Notez que les strings formattées ne sont pas un nouveau type de données, ce sont aussi des str!

```python
fstr = f"Compte numéro: {num + 12}; propriétaire: {name}; solde: {solde}"
print(fstr)
#ljust()
print(f"{'$'*10}{message}")
#rjust
print(f"{message}{'$'*10}")
# center
print(f"{'$'*10}{message}{'$'*10}")
```
