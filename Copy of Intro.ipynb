## Notion de basse des mots clés de la programmation

**1. C'est quoi la programmation ?**  
La programmation est l'action de concevoir et d’écrire des **instructions** (code) que va exécuter un ordinateur pour réaliser une tâche spécifique.

---

**2. C'est quoi un langage de programmation compilé ?**  
Un langage dont le code source est **traduit en une seule fois** en langage machine (fichier exécutable) avant son exécution.  
*Exemples :* C, C++, Rust, Go.

---

**3. C'est quoi un compilateur ?**  
Un **programme** qui lit tout le code source et le traduit en un fichier exécutable en langage machine, tout en vérifiant les erreurs de syntaxe on les retrouves très souvent avec les langagés bas niveau.

---

**4. C'est quoi un langage de programmation interprété ?**  
Un langage dont le code source est **exécuté ligne par ligne** par un interpréteur au moment de l’exécution, sans traduction préalable.  
*Exemples :* Python, JavaScript, Ruby, PHP.

---

**5. Qu'est-ce qu'un interpréteur ?**  
Un **programme** qui lit et exécute directement les instructions du code source, une par une, sans créer de fichier exécutable.

---

**6. C'est quoi un langage de programmation typé ?**  
Un langage où chaque **variable ou donnée a un type fixe** (entier, chaîne, booléen…).  
- *Typage statique* (vérifié à la compilation) : C, Java, Rust.  
- *Typage dynamique* (vérifié à l’exécution) : Python, JavaScript.

---

**7. C'est quoi un langage de programmation non typé ?**  
En pratique, il n’existe pas de langage vraiment « non typé ». On parle plutôt de **langage à typage faible** (conversions automatiques, ex. JavaScript avec `"5" * 2` donne `10`). Le contraire est un langage typé (cf. Q6).

---

**8. C'est quoi un algorithme ?**  
Une **suite finie et ordonnée d’étapes** permettant de résoudre un problème ou d’effectuer une tâche. C’est le plan logique avant la programmation.

---

**9. C'est quoi un langage de haut-niveau ? (optionnelle)**  
Un langage proche du langage humain, abstrait des détails matériels, lisible et portable.  
*Exemples :* Python, Java, C#, JavaScript.

---

**10. C'est quoi un langage de bas niveau ? (optionnelle)**  
Un langage proche du langage machine, donnant un contrôle fin sur le matériel, mais difficile à lire.  
*Exemples :* Assembleur, C (parfois dit « moyen niveau »).

---

**11. C'est quoi les entrées/sorties dans un programme ? (optionnelle)**  
- **Entrées** : données reçues de l'extérieur (clavier, fichier, réseau).  
- **Sorties** : résultats renvoyés (écran, fichier, imprimante).  
*Exemple :* `input()` / `print()` en Python.

---

**Récap' rapide :**  
- **Compilé** → traduction avant exécution (plus rapide).  
- **Interprété** → traduction pendant l’exécution (plus flexible).  
- **Typé** → types connus et contrôlés.  
- **Algorithme** → la logique ; **code** → l’implémentation.

## Les variables!
Attendez, il y a un truc qui va pas dans le programme précedent, **où** est donc passé le chiffre que vous avez donné après l'exécution de la cellule. C'est là le problème, il a tout simplement été perdu car on ne l'a pas affecté à une variable. Mais donc c'est quoi une variable? **Alright, here we go!**<br/>
Vous pouvez pensez aux variable comme des recipients pour stocker des trucs qu'on va pouvoir modifier, afficher ou supprimer autant qu'on veut dans notre programme. En python pour créer une variable il suffit de mettre le nom de cette dernière suivi du signe égal (=) et de l'élément que l'on veut mettre dans la variable, cela s'appelle une affectation. Exemple:
```
var = 10
```
Numéro 10 Leo Messi 😂, non je plaisant CR7 est plus inspirant mais Messi est meilleur <br/>
Cette syntax est la conséquence du fait que python n'est pas un langage typé, sinon il faudrait ajouter le type de la variable. Exemple avec le langage C:
```
int var; // Declaration
var = 10; // Affectation
// ou alors
int var = 10; // Declaration et affectation
```
Le mot clé **int** signifie **integer** qui veut dire entier en Français, avec ce genre de langage appélé langage à fort typage on doit préciser le type d'une variable à chaque fois qu'on la crée et en plus si on veut changer son contenu après on ne peut y mettre que des éléments qui ont le même type, c'est-à-dire par exemple que **var** ne pourrait contenir que des entiers dans le futur. Heureusement python nous épargne ça 🥳. Dans la cellule suivante on va sauver le nombre que vous donnerez moussallion!


````python
var = 10 # Crée la variable du nom de var contenant le nombre 10
print(var) # Affiche "10"
var = input("Donnez moi un nombre entre 0 et 100: ") # Change le contenu de var avec ce que vous allez donner
print("Vous avez donnez le nombre: ", var) # Affiche le nombre réçu
print("Le type de la variable var est: ", type(var)) # Affiche le type de la variable var, ici ce sera <class 'str'> car input() retourne toujours une chaine de caractère
````

## Mais alors il y a des types en python?
**Bien sûr que si**, une erreur que les débutants font souvent est de penser que les types n'esxitent pas dans les langages non typés. Détrompez-vous, ils existent mais c'est juste que grâce à la dynamicité de l'interpreteur, nous programmers n'auront pas à le preciser nous même. L'interpreteur le fera à notre place. Il existent énormément de type de données en python, c'est un des langages qui en a le plus même. On a des entiers (**int**), des réels (**float**), des chaînes de caractères (str) et bien d'autres, vous pourrez utiliser la classe **type** comme callable si vous voulez connaître le type d'une variable.<br/>
**Class**? **Callable**? ne vous inquietez pas on parlera de tout ça un peu plus tard

````python
# Comme vous le voyez les variables subsiste entre les cellules ce qui veut dire
# que vous pouvez créer une variable dans une cellule et l'utiliser dans une autre
print(var, type(var)) # Affichera le contenu de var et "<class str>"
var =10.0        # Changer le contenu de var pour y mettre 10.0
print(var, type(var)) # Affichera le contenu de var et "<class float>"
var = 10              # Changer le contenu de var pour y mettre 10 sans virgule
print(var, type(var)) # Affichera le contenu de var et "<class int>"
````

## Dynamicité des variables!
Comme vous le constatez on peut mettre n'importe quoi dans la même variable sans probème merci **Python** 😌. **Effectivement moussaillon**, tu as raison. La raison de cela est que en programmation les functions ont ce que l'on des **types de retour** qui sont les types de données retourné par une fonction et même des **type de paramatres** qui sont les types des données qu'on donne à la fonction. Mais bon on parlera de tout ça quand on fera la session sur les fonctions. ## Operateurs arithmétiques!
Ok donc dans un programme on peut créer et afficher des variables et recupérer des entrées, mais c'est tout?<br/>
**Bien sûr que non**! En général dans un programme on manipule les variables de tout type n'importe comment et comme on veut. Donc naturellement si on prend le cas particulier des nombres, on peut faire toutes les opérations qu'on fait dans la vrai vie avec des nombre comme l'**addition** (+), la **soustraction** (-), la **division** (/), la **multiplication** (* )<br/>
et le **reste de la division entière** (%). Ces derniers sont des operateurs basiques que vous retrouverez généralement avec les même symboles dans la plupart des langages de programmation par convention. Mais c'est pas tout car python fournit d'autre operateurs arithmétiques comme la **division entière** (//) et la **puissance** (**)

````python
#print(var,type(var)) # Affichera le contenu de var et "<class int>"
print(var % 2) # Affichera le reste de division de 77 par 2
print(var // 2) # Affichera le quotien de la division entière de 77 par 2
var2 = var**2 # Créer une nouvelle variable pour stocker 77 au carré
print(var2)     # Affichera le contenu de cette variable
````

## Operateurs d'incrémentationet d'affectation!
**Python** offre également une large pallete d'operateurs d'incrémentation un pour chaque operateur arithémetique enfaîte et même d'autres encore, on sait déja que le signe "=" fait l'affectation donc laissez moi vous présenter **(+=)** qui affecte en incrémentant. Par exemple:
```
var += 1 # equivaut à: var = var + 1
# et pareil pour toute les opération arithmétiques vu précedemment

```
On utilisera beaucoup ces operateurs dans les boucles **tant que** donc on y reviendra dans les prochains chapitres. Ce qu'il faut retenir c'est que incrémenter un variable c'est changer sa valeur d'un certain pas par rapport à elle même. Ok?


````python
var *= 2 # var = var * 2
print(var) # Comme vous le constatez la valeur de var à changer!
````

## Operateur logiques et booléens!
**Python** possède aussi des opérateurs logique, ces derniers retourne ce qu'on appelle des booléens c'est-à-dire soit **Vrai** soit **Faux** (**True/False**). **Attention** ces valeur là ne sont pas des variables, on en parlera un peu plus en détail quand on parlera des types de données. Pour le moment ce que l'on doit savoir c'est que les opérateurs logique retournent ce types de valeur et on verra ce qu'on fait avec des booléens quand on parlera de condition plus tard dans ce notebook.<br/>

Plein d'autres opérateurs retournent des booléens tels que les **opérateurs de comparaison** qu'on verra dans la prochaine section mais également des **operateurs de contenance** qu'on verra un peu plus tard dans le programme et des **operateurs de manipulation de Bytes** et d'autres encore.

````python
boolean = True
print(boolean is True) # Perform l'operation d'identité avec variable boolean, Attention ceci est différent de la comparaison d'égalité
print(boolean or False) # Le ET logic
print(True + True + 1) # Qu'est ce que ça veut selon vous?
````

## Les conditions!
Ok on a des booléens mais il servent à quoi concretement?<br/>
Très bien voilà qui nous amène à la dernière section de ce chapitre les **structures conditionnelles**. Pour écrire des programmes capable de s'adapter en cas d'incerticude les structures conditionnelles sont indispensable, c'est pour ils existent dans tous les langages de programmation, parce que dans la vraie vie on ne contrôle pas toujours tout. Les **structures conditionnelles** ne sont rien de plus que des closes qu'on utilise pour que notre programme puisse avoir quoi faire dans différents scenarios. C'est comme dans la vie de tous les jours les "**si**", "**sinon si**" et "**sinon**" (**if, elif, else**) et devinez quoi?<br/>
**Oui exactement**, elles marchent avec des booléens. Une mot clés d'une structure conditionnelle comme le **if** par exemple évalue une condition donnée et attend que cette évaluation lui retourne une valeur de vérité soit **True** soit **False**. Avant d'executer un bout de code appélé **block**. Ce block doit être un peu espacé par rapport à la ligne de la condition, on appelle ça l'indentation et python est très stricte sur ça, c'est son **péché mignon** 😂.<br/>
En gros les structures condionnelles ce sont juste des **blocks de code** écrit de manière à ce qu'il ne soit exécuté que si une certaine condition est evalué comme étant vrai:
```
if taille < 170:
  print("Vous êtes court")
elif taille > 180:
  print("Vous êtes grand")
else:
  print("Vous êtes un homme de taille normal")
```

````python
if var < var2:
  print(var, "est inferieur à ", var2)

if True:
  print("Ce texte s'affichera toujours car la condition est toujours vrai!")
````

Et bien que en programmation on evitera cela la plupart du temps, **python** a une strategie super pour ce cas. En python toute condition qui n'est pas explicitement fausse est considérer comme vrai. Check this out!

# Formule de résolution d'une équation du second degré

## 📐 Forme générale

Une équation du second degré s'écrit sous la forme :

$$
\Large ax^2 + bx + c = 0
$$

Avec :
- $a \neq 0$ (sinon ce n'est plus une équation du second degré)
- $a$, $b$, $c$ sont des **coefficients réels** (ou complexes)

---

## 🧮 Formule de résolution (discriminant)

### 1. Calcul du discriminant

$$
\Large \Delta = b^2 - 4ac
$$

---

### 2. Tableau complet des solutions

| Cas | Valeur de $\Delta$ | Nature des solutions | Formule des solutions |
|-----|-------------------|----------------------|----------------------|
| **Cas 1** | $\Delta > 0$ | Deux solutions **réelles distinctes** | $$x_1 = \frac{-b - \sqrt{\Delta}}{2a}$$ <br> $$x_2 = \frac{-b + \sqrt{\Delta}}{2a}$$ |
| **Cas 2** | $\Delta = 0$ | Une solution **réelle double** | $$x_0 = \frac{-b}{2a}$$ |
| **Cas 3** | $\Delta < 0$ | Deux solutions **complexes conjuguées** | $$x_1 = \frac{-b - i\sqrt{-\Delta}}{2a}$$ <br> $$x_2 = \frac{-b + i\sqrt{-\Delta}}{2a}$$ |

> ⚠️ **Attention :** Pour $\Delta < 0$, on utilise $\sqrt{-\Delta}$ (avec $-\Delta > 0$) et non $\sqrt{\Delta}$ qui n'est pas défini dans les réels.

---

### 3. Exemples détaillés

#### Exemple 1 : $\Delta > 0$

Résoudre $2x^2 - 3x - 2 = 0$

| Étape | Calcul |
|-------|--------|
| Coefficients | $a = 2$, $b = -3$, $c = -2$ |
| Discriminant | $\Delta = (-3)^2 - 4 \times 2 \times (-2) = 9 + 16 = 25$ |
| $\sqrt{\Delta}$ | $\sqrt{25} = 5$ |
| Solutions | $x_1 = \dfrac{3 - 5}{4} = -\dfrac{1}{2}$ <br> $x_2 = \dfrac{3 + 5}{4} = 2$ |
| **Résultat** | $\boxed{x_1 = -\dfrac{1}{2}}$ et $\boxed{x_2 = 2}$ |

---

#### Exemple 2 : $\Delta = 0$

Résoudre $x^2 - 6x + 9 = 0$

| Étape | Calcul |
|-------|--------|
| Coefficients | $a = 1$, $b = -6$, $c = 9$ |
| Discriminant | $\Delta = (-6)^2 - 4 \times 1 \times 9 = 36 - 36 = 0$ |
| Solution | $x_0 = \dfrac{6}{2} = 3$ |
| **Résultat** | $\boxed{x = 3 \text{ (solution double)}}$ |

---

#### Exemple 3 : $\Delta < 0$ (avec complexes)

Résoudre $x^2 + x + 1 = 0$

| Étape | Calcul |
|-------|--------|
| Coefficients | $a = 1$, $b = 1$, $c = 1$ |
| Discriminant | $\Delta = 1^2 - 4 \times 1 \times 1 = 1 - 4 = -3$ |
| $-\Delta$ | $-\Delta = 3$ |
| $\sqrt{-\Delta}$ | $\sqrt{3}$ |
| Solutions | $x_1 = \dfrac{-1 - i\sqrt{3}}{2}$ <br> $x_2 = \dfrac{-1 + i\sqrt{3}}{2}$ |
| **Résultat** | $\boxed{x_1 = -\dfrac{1}{2} - i\dfrac{\sqrt{3}}{2}}$ <br> $\boxed{x_2 = -\dfrac{1}{2} + i\dfrac{\sqrt{3}}{2}}$ |

---

### 4. Exemple avec vérification

Vérifions que $x_1 = -\dfrac{1}{2} - i\dfrac{\sqrt{3}}{2}$ est bien solution de $x^2 + x + 1 = 0$ :

| Étape | Calcul |
|-------|--------|
| Calcul de $x_1^2$ | $\left(-\dfrac{1}{2} - i\dfrac{\sqrt{3}}{2}\right)^2 = \dfrac{1}{4} + i\dfrac{\sqrt{3}}{2} - \dfrac{3}{4} = -\dfrac{1}{2} + i\dfrac{\sqrt{3}}{2}$ |
| Calcul de $x_1^2 + x_1 + 1$ | $\left(-\dfrac{1}{2} + i\dfrac{\sqrt{3}}{2}\right) + \left(-\dfrac{1}{2} - i\dfrac{\sqrt{3}}{2}\right) + 1 = 0$ |
| **Conclusion** | $\boxed{x_1 \text{ est bien solution !}}$ |

---

## 📊 Tableau des propriétés

| Propriété | Formule | Condition |
|-----------|---------|-----------|
| **Discriminant** | $\Delta = b^2 - 4ac$ | Toujours valable |
| **Somme des racines** | $S = x_1 + x_2 = -\dfrac{b}{a}$ | $\Delta \geq 0$ |
| **Produit des racines** | $P = x_1 \times x_2 = \dfrac{c}{a}$ | $\Delta \geq 0$ |
| **Forme factorisée** | $ax^2 + bx + c = a(x - x_1)(x - x_2)$ | $\Delta \geq 0$ |
| **Sommet de la parabole** | $x_S = -\dfrac{b}{2a}$ <br> $y_S = -\dfrac{\Delta}{4a}$ | Toujours valable |

---

## 📈 Lien entre $\Delta$ et la parabole

| Valeur de $\Delta$ | Graphique | Intersection avec l'axe des $x$ |
|--------------------|-----------|--------------------------------|
| $\Delta > 0$ | Parabole coupant l'axe en **deux points** | $x_1$ et $x_2$ |
| $\Delta = 0$ | Parabole **tangente** à l'axe des $x$ | Un seul point $x_0$ |
| $\Delta < 0$ | Parabole **ne coupant pas** l'axe des $x$ | Pas d'intersection réelle |


````python
## Votre code ici!*
print("Bienvenu dans votre calculateur de racine carré de la fonction ax^2+bx+c=0")
#recupeartion de la saisie de l'utilisateur pour prendre toutes les coeficients de la fonction ax^2+bx+c=0
a = float(input("Donner a: "))
b = float(input("Donner b: "))
c = float(input("Donner c: "))


delta = b**2 - 4*a*c
if delta < 0:
  print("Il n'y a pas de solution réelle")
elif delta == 0:
  x = -b/(2*a)
  print("Il y a une solution réelle: ", x)
else:
  x1 = (-b + delta**0.5)/(2*a)
  x2 = (-b - delta**0.5)/(2*a)
  print("Il y a deux solutions réelles: ", x1, " et ", x2)
````

## Switch case in python! (optionnelle)
Donc il y'a un autre concept pas si important dont je tenais à parler. C'est les statements **switch/case** qui existent dans énormement de langages de programmation mais qui n'avait pas équivalent python jusqu'à la version 3.10 de python justement. Mais pour être complet quand on parle de structures conditionnelle dans la programmation en générale, il faut en parler!<br/>
Pour faire simple les switch/case sont juste une autre manière de faire des stuctures conditionnelle optimisé pour les cas la condition est basée sur une variable à valeur discrète et finie. Par exemple, supposons qu'on a une variable qui peut prendre que trois valeurs: "jour", "soir" et "nuit". basiquement un switch/case statement sert à ce genre de cas, on peut évidemment faire ça avec des if et else if aussi mais le dans la plupart des langages le switch/case est codé et compilé pour contenir moins d'instruction que les structures classiques avec ce genre de cas, donc il sera plus efficace algorithmiquement parlant mais est-ce aussi cas en python? Question! 😂 Même si certains affirment qu'il serait environ **86**% plus efficace ça reste relatif. On parlera encore de ça surement si on le temps de parlé de complexité algorithmique. Mais en bref voilà comment ça présenterait en C:

```
switch (expression)
​{
    case constant1:
      // statements
      break;

    case constant2:
      // statements
      break;
    .
    .
    .
    default:
      // default statements
}
```
En python depuis la version 3.10, le statement à un équivalent le **match/case**, même si beaucoup se demandent les dévéloppeurs n'ont pas juste fait pour répondre au démandes des pythonistas car contrairement à tous les autres mots clés python **match** et **case** sont toujous utilisable comme nom de variable, donc ils sont en quelque sorte des **pseudo keywords** mais **never mind, ça fait notre affaire**

````python
time = "jou4r"
match time:
  case "jour":
    print("Bonjour")
  case "soir":
    print("Bonsoir")
  case "nuit":
    print("Bonne nuit")
  case _:   # En python le default case correspont à underscore
    print("???")
````
